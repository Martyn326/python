### 1️⃣ Suma de los números del 1 al 50 que sean múltiplos de 5
**Problema:** Escribe una función `suma_multiplos_5()` que calcule la suma de todos los múltiplos de 5 entre 1 y 50 (ambos incluidos).

```python
def suma_multiplos_5():
    suma_multiplos = 0
    for n in range(1,51):
        if n % 5 == 0:
            suma_multiplos += n
    return suma_multiplos
```

```python
assert suma_multiplos_5() == 275
```

**Problema:** Implementa una función `contar_impares(inicio, fin)` que cuente cuántos números impares hay entre los valores dados (inclusive).

```python
def contar_impares(inicio, fin):
    contar_impares = 0 
    for i in range(inicio, fin + 1):
        if i % 2 != 0:
            contar_impares += 1
    return contar_impares
    
```

### 4️⃣ Buscar la primera vocal en una palabra usando `break`
**Problema:** Crea `primera_vocal(palabra)` que devuelva la primera vocal encontrada en la palabra. Si no hay vocales, devuelve `None`.

```python
def primera_vocal(palabra):
  vocal = None
  vocales = "aeiouAEIOU"
  for n in palabra:
    if n in vocales:
      vocal = n
      break
  return vocal
```

```python
assert primera_vocal('python') == 'o'
assert primera_vocal('sky') == None
```
